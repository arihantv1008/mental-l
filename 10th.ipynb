Develop a program to implement k-means clustering using Wisconsin Breast Cancer data set and visualize 
the clustering result.

```python
from sklearn.datasets import load_breast_cancer
```

```python
cancer = load_breast_cancer()
```

```python
from sklearn.cluster import KMeans
predict = KMeans(n_clusters=2).fit_predict(cancer.data)
```

```python
import matplotlib.pyplot as plt
plt.scatter(cancer.data[:,0],cancer.data[:,1],c=predict)
```
