Develop a program to implement the Naive Bayesian classifier considering Olivetti Face Data set for training. 
Compute the accuracy of the classifier, considering a few test data sets. 

```python
from sklearn.datasets import fetch_olivetti_faces
```

```python
faces = fetch_olivetti_faces()
faces.keys()
```

```python
from sklearn.model_selection import train_test_split
```

```python
xtrain,xtest,ytrain,ytest, imgtrain,imgtest = train_test_split(faces.data, faces.target, faces.images, test_size=0.2, random_state=42)
```

```python
from sklearn.naive_bayes import GaussianNB
```

```python
prediction = GaussianNB().fit(xtrain,ytrain).predict(xtest)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(20,15))
for i in range(10):
    plt.subplot(5,5,i+1)
    plt.imshow(imgtest[i])
    plt.title(f"Predicted:{prediction[i]}\nOriginal:{ytest[i]}")
plt.tight_layout()
plt.show()
```
