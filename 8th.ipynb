Develop a program to demonstrate the working of the decision tree algorithm. Use Breast Cancer Data set 
for building the decision tree and apply this knowledge to classify a new sample.

```python
from sklearn.datasets import load_breast_cancer
```

```python
cancer = load_breast_cancer()
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest, ytrain, ytest = train_test_split(cancer.data, cancer.target, test_size=0.2, random_state=42)
```

```python
from sklearn.tree import DecisionTreeClassifier, plot_tree
```

```python
tree = DecisionTreeClassifier(criterion="gini").fit(xtrain,ytrain)
import matplotlib.pyplot as plt
plt.figure(figsize=(20,25))
plot_tree(tree,filled=True)
plt.show()
```
