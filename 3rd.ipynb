Develop a program to implement Principal Component Analysis (PCA) for reducing the dimensionality of 
the Iris dataset from 4 features to 2.

```python
from sklearn.datasets import load_iris
```

```python
iris = load_iris()
data = iris.data
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
scaler = StandardScaler().fit_transform(data)
pc = PCA(n_components=2).fit_transform(scaler)
```

```python
import pandas as pd

df = pd.DataFrame(pc,columns=["pc1","pc2"])
df["Target"] = iris.target
```

```python
import seaborn as sns
sns.scatterplot(x="pc1",y="pc2", hue="Target", data=df)
```
