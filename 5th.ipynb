Develop a program to implement k-Nearest Neighbour algorithm to classify the randomly generated 100 
values of x in the range of [0,1]. Perform the following based on dataset generated. 
a. Label the first 50 points {x1,……,x50} as follows: if (xi ≤ 0.5), then xi ∊ Class1, else xi ∊ Class1 
b. Classify the remaining points, x51,……,x100 using KNN. Perform this for k=1,2,3,4,5,20,30

```python
import numpy as np 
```

```python
x = np.random.rand(100).reshape(-1,1)
y = np.array([1 if val <= 0.5 else 2 for val in x[:50]])
```

```python
xtrain = x[:50]
ytrain = y
xtest = x[50:]
```

```python
kValues = [1,2,3,20,40]
import matplotlib.pyplot as plt
from sklearn.neighbors import KNeighborsClassifier
```

```python
plt.figure(figsize=(20,15))
for i, value in enumerate(kValues, 1):
    predicted = KNeighborsClassifier(n_neighbors=value).fit(xtrain,ytrain).predict(xtest)
    plt.subplot(5,2,i)
    plt.scatter(xtrain, ytrain, c="red", label="Tested")
    plt.scatter(xtest, predicted, c="blue", label="Predicted")
    plt.legend()
    plt.yticks(predicted)    
plt.tight_layout()
plt.show()
```
